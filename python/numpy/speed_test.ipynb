```python
#using python list
import numpy as np
import time

s = 1_000_000
l1 = list(range(s))

l2=list(range(s))
start = time.time()
add = [ x+y for x,y in zip(l1,l2)]
print(add[0:10])
end = time.time()
print(end-start)
```

```python
# using numpy array
import numpy as np
import time

s = 1_000_000
l1 = np.array(list(range(s)))

l2=np.array(list(range(s)))
start = time.time()
add = l1+l2
print(add[0:10])
end = time.time()
print(end-start)
```
