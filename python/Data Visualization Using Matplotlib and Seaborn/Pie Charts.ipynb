```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")
```

```python
# Data
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels)
plt.show()
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs,labels=labels,colors=colors)
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={"edgecolor":'#050407' ,"linewidth":5,"linestyle":(0, (3, 1, 1, 1, 1, 1))}
)
```

```python
explode=[0.1,0,0,0]
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    explode=explode,
    wedgeprops={"edgecolor":'#050407' ,"linewidth":5,"linestyle":(0, (3, 1, 1, 1, 1, 1))}
)
plt.title("Exploded pie example")
```

```python
explode=[0.1,0,0,0]
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    explode=explode,
    shadow=True,
    startangle=100
)
```

```python
explode=[0.1,0,0,0]
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    explode=explode,
    shadow=True,
    startangle=100,
    autopct='%1.1f%%'
)
plt.title("percent pie")
```

```python
# Too many categories
languages = ['Python', 'Java', 'C++', 'JavaScript', 'C#', 'Ruby', 'Go', 'Rust', 'Swift', 'PHP']
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]

plt.pie(usage, labels=languages, autopct='%1.0f%%', startangle=90)
plt.title("Programming Language Usage (Crowded Example)")
plt.show()
plt.figure(figsize=(6,6))
plt.tight_layout()
```
