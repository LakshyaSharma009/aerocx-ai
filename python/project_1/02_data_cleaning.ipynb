```python
#Load the data
import json

def clean_data(data):
    #removing rows with no names
    data["users"]=[user for user in data["users"] if user["name"].strip()]
    #removing duplicate elements in the friend list
    for user in data["users"]:
        user["friends"]=list(set(user["friends"]))
    #removing inactive users
    data["users"]=[ user for user in data["users"] if user["friends"] or user["liked_pages"]]

    #removing duplicate pages
    unique_pages={}
    for page in data["pages"]:
        unique_pages[page["id"]]=page
    data["pages"]=list(unique_pages.values())
    return data
    


data=json.load(open("massive_data.json"))
data=clean_data(data)

json.dump(data,open("cleaned_massive_data.json","w"),indent=4)
print("Data has been cleaned Succcessfully")
```
